```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = r'downloads\job_posting_clean.csv' 
df = pd.read_csv(file_path)

df = df.dropna(subset=['max_salary', 'med_salary', 'min_salary', 'formatted_experience_level'])

experience_levels = df.groupby('formatted_experience_level').agg({
    'max_salary': 'mean',
    'med_salary': 'mean',
    'min_salary': 'mean'
}).reset_index()

plt.figure(figsize=(10, 6))
plt.bar(experience_levels['formatted_experience_level'], experience_levels['max_salary'], label='Max Salary')
plt.bar(experience_levels['formatted_experience_level'], experience_levels['med_salary'], label='Median Salary')
plt.bar(experience_levels['formatted_experience_level'], experience_levels['min_salary'], label='Min Salary')

plt.xlabel('Experience Level')
plt.ylabel('Average Salary (USD)')
plt.title('Salary Differences Across Experience Levels')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
